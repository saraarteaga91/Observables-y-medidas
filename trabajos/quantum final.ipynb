```python
#Puntos 1 y 2.
import numpy as np
estado_a = np.array([complex(-3, -1), complex(0, -2), complex(0, 1), complex(2, 0)])
estado_b = np.array([complex(1, 1), complex(0, 1), complex(1, -1), complex(0, 2)])
def obtener_probabilidad_en_indice(estado_cuantico, indice_cercano):
    total_norma = np.sum(np.abs(estado_cuantico)**2)
    return np.abs(estado_cuantico[indice_cercano])**2 / total_norma
def calcular_probabilidad_cambio(estado_a, estado_b):
    producto_intermedio = np.vdot(estado_a, estado_b)  
    return np.abs(producto_intermedio)**2
probabilidad_indice_2 = obtener_probabilidad_en_indice(estado_a, 2)
print(f"Probabilidad en el índice 2: {probabilidad_indice_2}")
probabilidad_transicion_resultante = calcular_probabilidad_cambio(estado_a, estado_b)
print(f"Probabilidad de transición: {probabilidad_transicion_resultante}")
```

```python
#COMPLETE LOS RETOS DE PROGRAMACIÓN DEL CAPÍTULO 4.
#1. Amplitud de transición. El sistema puede recibir dos vectores y calcular la probabilidad de transitar de el uno al otro después de hacer la observación
import numpy as np
def transicion(vec1, vec2):
    return np.abs(np.dot(vec1/np.linalg.norm(vec1), vec2/np.linalg.norm(vec2)))**2
vec1 = np.array([0.5, 1/3])
vec2 = np.array([1, 1])
print(f"Probabilidad de transición: {transicion(vec1, vec2)}")
```

```python
#2. Ahora con una matriz que describa un observable y un vector ket, el sistema revisa que la matriz sea hermitiana, y si lo es, calcula la media y la varianza del observable en el estado dado.
import numpy as np
def es_hermitiana(matriz):
    return np.allclose(matriz, matriz.T.conj())
def calcular_media_y_varianza(obs, ket):
    media = np.vdot(ket, np.dot(obs, ket)).real
    varianza = np.vdot(ket, np.dot(obs @ obs, ket)).real - media**2
    return media, varianza
obs = np.array([[1, -1j], [1j, 2]])
ket = np.array([np.sqrt(2)/2, np.sqrt(2)/2j])
if es_hermitiana(obs):
    media, varianza = calcular_media_y_varianza(obs, ket)
    print(f"Media: {media}")
    print(f"Varianza: {varianza}")
else:
    print("La matriz no es hermitiana.")
```

```python
#3. El sistema calcula los valores propios del observable y la probabilidad de que el sistema transite a alguno de los vectores propios después de la observación.
import numpy as np
def obtener_vectores_propios(obs):
    valores_propios, vectores_propios = np.linalg.eig(obs)
    print("Valores propios:", valores_propios)
    return vectores_propios
def calcular_probabilidad(vectores_propios, ket):
    ket_norm = ket / np.linalg.norm(ket)
    proyecciones = np.abs(np.dot(vectores_propios.T.conj(), ket_norm)) ** 2
    return proyecciones
obs = np.array([[-1, -1j], [1j, 1]])
ket = np.array([1/2, 1/2]).reshape(-1, 1)
vectores_propios = obtener_vectores_propios(obs)
probabilidad = calcular_probabilidad(vectores_propios, ket)
print("Probabilidad de transición:", probabilidad)
```

```python
#4. Se considera la dinámica del sistema. Ahora con una serie de matrices Un el sistema calcula el estado final a partir de un estado inicial.
import numpy as np
def calcular_estado(estado_inicial, matrices):
    for matriz in matrices:
        estado_inicial = matriz @ estado_inicial
    return estado_inicial
estado_inicial = np.array([1, 0])
matrices = [np.eye(2), np.array([[0, 1], [1, 0]])]  
estado_final = calcular_estado(estado_inicial, matrices)
estado_final = estado_final.astype(int) 
print("Estado final:", estado_final)
```

```python
#REALICE LOS SIGUIENTES PROBLEMAS E INCLUYALOS COMO EJEMPLOS
#4.3.1
import numpy as np
Sx = (1/np.sqrt(2)) * np.array([[0, 1], [1, 0]])
estado_inicial = np.array([[1], [0]])
estado_final = Sx @ estado_inicial
probabilidad_up = abs(np.dot([1, 0], estado_final)) ** 2
probabilidad_down = abs(np.dot([0, 1], estado_final)) ** 2
print(f"Probabilidad spin_up: {probabilidad_up}")
print(f"Probabilidad spin_down: {probabilidad_down}")
```

```python
#4.3.2
import numpy as np
import matplotlib.pyplot as plt
observable = np.array([[1, 0], [0, 2]], dtype=complex)
estado_inicial = np.array([1, 1], dtype=complex) / np.sqrt(2)
valores, vectores = np.linalg.eigh(observable)
probs = np.abs(vectores.T @ estado_inicial)**2
plt.bar(valores, probs)
plt.xlabel('Eigenvalues')
plt.ylabel('Transition Probability')
plt.title('Probability Distribution')
plt.show()
```

```python
#4.5.2
import numpy as np
up = np.array([1, 0])    # |↑>
down = np.array([0, 1])  # |↓>
def prod_tensorial(estado1, estado2):
    return np.kron(estado1, estado2)
# Crear el estado de Bell |ψ> = (|↑↓> + |↓↑>)/√2
estado_bell = (prod_tensorial(up, down) + prod_tensorial(down, up)) / np.sqrt(2)
print("Estado entrelazado de Bell |ψ>:\n", estado_bell)
```
